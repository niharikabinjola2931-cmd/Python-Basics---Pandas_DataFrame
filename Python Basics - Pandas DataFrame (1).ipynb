## Question 1 :  Write a Python program to check whether a number is even or odd using conditional statements.

```python
num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

## Question 2 : Write a Python program to count how many times a number appears in a list.

Example List

[2,3,4,2,5,2]

```python
numbers = [2, 3, 4, 2, 5, 2]

num = int(input("Enter the number to count: "))

count = numbers.count(num)

print("The number appears", count, "times")
```

## Question 9 : Create a random array of 10 numbers and:

Find max and min

Sort the array

```python
import numpy as np

arr = np.random.randint(1, 100, 10)

print("Array:", arr)
print("Maximum:", arr.max())
print("Minimum:", arr.min())
print("Sorted array:", np.sort(arr))
```
